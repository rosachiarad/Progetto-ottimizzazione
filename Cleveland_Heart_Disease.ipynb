```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from ucimlrepo import fetch_ucirepo
from sklearn.model_selection import train_test_split,StratifiedKFold
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.metrics import confusion_matrix,classification_report,roc_curve,auc

#---------------------------------------------------
# 1. EDA - Heart Disease Cleveland
#---------------------------------------------------
heart_disease=fetch_ucirepo(id=45)
X=heart_disease.data.features
y=heart_disease.data.targets

df=pd.concat([X,y],axis=1)
print("Dataset Cleveland Heart Disease")
display(df.head())
print("Descrizione")
display(df.describe())

target_col=y.columns[0]
df['target_binary']=df[target_col].apply(lambda x:1 if x>0 else 0)

print("\nValori nulli nel dataset:")
print(df.isnull().sum()[df.isnull().sum()>0])
print(f"\nRighe totali mantenute per l'addestramento:{df.shape[0]}")

colonne_da_rimuovere=[target_col,'target_binary']
X=df.drop(columns=colonne_da_rimuovere,errors='ignore')
y=df['target_binary']

sns.set_theme(style="ticks", palette="muted")
#GRAFICO 1: Confronto Distribuzione Target 
plt.figure(figsize=(12, 5))
ax1 = plt.subplot(1, 2, 1)
sns.countplot(
    data=df, x=target_col, ax=ax1, 
    hue=target_col, palette='Set2', legend=False, edgecolor='black'
)
for container in ax1.containers:
    ax1.bar_label(container, label_type='edge', fontweight='bold', padding=3)
plt.title(f"Distribuzione target originale (0-4)", fontsize=14, fontweight='bold', pad=10)
plt.ylabel("numero Pazienti", fontweight='bold')
plt.ylim(0, 190) 


ax2 = plt.subplot(1, 2, 2)
sns.countplot(
    data=df, x='target_binary', ax=ax2, 
    hue='target_binary', palette='Set2', legend=False, edgecolor='black'
)
for container in ax2.containers:
    ax2.bar_label(container, label_type='edge', fontweight='bold', padding=3)
plt.title("Distribuzione delle classi", fontsize=14, fontweight='bold', pad=10)
plt.ylabel("numero pazienti", fontweight='bold')
plt.ylim(0, 190)

sns.despine()
plt.tight_layout()
plt.show()

#GRAFICO 2: Distribuzione variabili continue 
num_cols = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak']
df[num_cols].hist(
    figsize=(14, 10), bins=20, edgecolor='black', 
    color='#1f77b4', grid=False
)
plt.suptitle("Istogrammi delle variabili continue", y=1.02, fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()

X_train_full, X_test, y_train_full, y_test = train_test_split(X,y,test_size=0.2,stratify=y,random_state=42)

# ------------------------------------------------------------------
# 1.1 PREPROCESSING
#  ------------------------------------------------------------------
num_features=['age','trestbps','thalach','oldpeak','chol','ca']
cat_features=['cp','restecg','slope','thal']

num_pipeline=Pipeline([
('imputer',SimpleImputer(strategy='median')),
('scaler',StandardScaler())
])

cat_pipeline=Pipeline([
('imputer',SimpleImputer(strategy='most_frequent')),
('encoder',OneHotEncoder(handle_unknown='ignore',sparse_output=False))
])

ct=ColumnTransformer(
transformers=[
('num',num_pipeline,num_features),
('cat',cat_pipeline,cat_features)
],
remainder='passthrough'
)


X_train_full_processed_=ct.fit_transform(X_train_full)
#estraggo i nomi puliti delle nuove colonne
nomi_feature_trasformate = [
    n.replace('num__', '').replace('cat__', '').replace('remainder__', '') 
    for n in ct.get_feature_names_out()
]
#creo il DataFrame per visualizzarlo
df_train_trasformato=pd.DataFrame(X_train_full_processed_, columns=nomi_feature_trasformate)
print("\n ANTEPRIMA DATASET DOPO IL PREPROCESSING")
display(df_train_trasformato.head())
# -------------------------------------------------------------------------------------------------
# 2. CLASSE REGRESSIONE LOGISTICA e METODO DI OTTIMIZZAZIONE GD, NEWTON, NEWTON CG, BFGS, L-BFGS
#---------------------------------------------------------------------------------------------------
class RegressioneLogistica:
    def __init__(self,metodo='gd',max_iter=1000,tol=1e-5,c1=1e-4,c2=0.9,lambda_reg=1.0,m=10):
        self.metodo=metodo.lower()
        self.max_iter=max_iter
        self.tol=tol
        self.c1=c1
        self.c2=c2
        self.lambda_reg=lambda_reg
        self.m=m
        self.w=None
        self.loss_storia=[]
        self.s_history=[]
        self.y_history=[]

    def sigmoide(self,z):
        return 1/(1+np.exp(-z))

    def calcola_gradiente_loss(self,w,X,y):
        N=len(y)
        ypred=self.sigmoide(np.dot(X,w))
        loss=-np.mean(y*np.log(ypred)+(1-y)*np.log(1-ypred))

        w_reg=np.copy(w)
        w_reg[0]=0
        loss+=(self.lambda_reg/(2*N))*np.sum(w_reg**2) #f

        grad=np.dot(X.T,(ypred-y))/N
        grad+=(self.lambda_reg/N)*w_reg
        return loss,grad,ypred

    def calcola_hessiana(self,w,X,ypred):
        N=X.shape[0]
        S_vec=ypred*(1-ypred)
        H=np.dot(X.T,X*S_vec[:,np.newaxis])/N

        I_reg=np.eye(X.shape[1])
        I_reg[0,0]=0
        H+=(self.lambda_reg/N)*I_reg
        return H

    def cg_newton(self,grad,B_k):
        grad_norm=np.linalg.norm(grad)
        epsilon_k = min(0.5,np.sqrt(grad_norm))*grad_norm
        z=np.zeros_like(grad)
        r=np.copy(grad)
        d=-np.copy(grad)
        r_norm_sq=np.dot(r,r)
        max_j=len(grad)*2

        for j in range(max_j):
            Bd=np.dot(B_k,d)
            curvatura=np.dot(d,Bd)

            if curvatura<=0:
                if j==0: return -grad
                else: return z

            alpha_j=r_norm_sq/curvatura
            z=z+alpha_j*d
            r_next=r+alpha_j*Bd

            if np.linalg.norm(r_next)<epsilon_k:
                return z

            r_next_norm_sq=np.dot(r_next,r_next)
            beta_next=r_next_norm_sq/r_norm_sq
            d=-r_next+beta_next*d
            r=r_next
            r_norm_sq=r_next_norm_sq
        return z

    def lbfgs(self,grad):
        q=np.copy(grad)
        m_hist=len(self.s_history)
        if m_hist==0: return -q

        alphas,rhos=np.zeros(m_hist),np.zeros(m_hist)
        for i in range(m_hist-1,-1,-1):
            s_i,y_i=self.s_history[i],self.y_history[i]
            rhos[i]=1.0/np.dot(y_i,s_i)
            alphas[i]=rhos[i]*np.dot(s_i,q)
            q-=alphas[i]*y_i

        s_last,y_last=self.s_history[-1],self.y_history[-1]
        gamma=np.dot(s_last,y_last)/np.dot(y_last,y_last) if np.dot(y_last,y_last)>1e-10 else 1.0
        z=gamma*q

        for i in range(m_hist):
            s_i,y_i=self.s_history[i],self.y_history[i]
            beta=rhos[i]*np.dot(y_i,z)
            z+=s_i*(alphas[i]-beta)
        return -z

    def backtracking_ls_armijo(self,w,X,y,direzione,loss_attuale,grad,iterazione,loss_precedente):
        deriv_dir=np.dot(grad,direzione)
        if self.metodo in ['newton','cgn']:
            alpha=1
        elif self.metodo=='gd':
            if iterazione==0: alpha=1/(np.linalg.norm(grad))
            else:
                alpha=2*(loss_attuale-loss_precedente)/deriv_dir
                alpha=min(1,1.01*alpha)

        rho=0.5
        alpha_min=1e-10
        w_new=w+alpha*direzione
        loss_new,_,_=self.calcola_gradiente_loss(w_new,X,y)

        while not (loss_new<loss_attuale+self.c1*alpha*deriv_dir):
            alpha*=rho
            w_new=w+alpha*direzione
            loss_new,_,_=self.calcola_gradiente_loss(w_new,X,y)

        return alpha

    #LS PER LE CONDIZIONI DI WOLFE (ALGORITMO 3.6 NOCEDAL)
    def ls_wolfe(self,w,X,y,direzione,loss_attuale,grad):
        alpha_max,alpha_prev,alpha=100,0,1
        deriv_dir_old=np.dot(grad,direzione)
        loss_prev=loss_attuale
        i,max_iter=1,50

        while i<max_iter:
            w_new=w+alpha*direzione
            loss_new,grad_new,_=self.calcola_gradiente_loss(w_new,X,y)

            if (loss_new>loss_attuale+self.c1*alpha*deriv_dir_old) or (loss_new>=loss_prev and i>1):
                return self.zoom(w,X,y,direzione,alpha_prev,alpha,loss_attuale,deriv_dir_old)

            deriv_dir_new=np.dot(grad_new,direzione)
            if abs(deriv_dir_new)<=-self.c2*abs(deriv_dir_old): return alpha
            if deriv_dir_new>=0:
                return self.zoom(w,X,y,direzione,alpha,alpha_prev,loss_attuale,deriv_dir_old)

            alpha_prev,loss_prev=alpha,loss_new
            alpha=min(alpha*2.0,alpha_max)
            i+=1
        return alpha

    #ZOOM
    def zoom(self,w,X,y,direzione,alpha_low,alpha_high,loss_attuale,deriv_dir_old):
        iterazione,max_iter=0,50
        while iterazione<max_iter:
            iterazione+=1
            alpha=0.5*(alpha_low+alpha_high)
            w_new=w+alpha*direzione
            loss_new,grad_new,_=self.calcola_gradiente_loss(w_new,X,y)

            w_low=w+alpha_low*direzione
            loss_low,_,_=self.calcola_gradiente_loss(w_low,X,y)

            if (loss_new>loss_attuale+self.c1*alpha*deriv_dir_old) or (loss_new>=loss_low):
                alpha_high=alpha
            else:
                deriv_dir_new=np.dot(grad_new,direzione)
                if abs(deriv_dir_new)<=-self.c2*deriv_dir_old: return alpha
                if deriv_dir_new*(alpha_high-alpha_low)>=0: alpha_high=alpha_low
                alpha_low=alpha
        return alpha

    def fit(self,X,y):
        X_bias=np.insert(X,0,1,axis=1)
        self.w=np.zeros(X_bias.shape[1])
        y=np.array(y).flatten()
        if self.metodo=='bfgs': H_inv=np.eye(X_bias.shape[1])

        self.s_history,self.y_history=[],[]
        loss_precedente=None

        for i in range(self.max_iter):
            loss,grad,ypred=self.calcola_gradiente_loss(self.w,X_bias,y)
            self.loss_storia.append(loss)
            if np.linalg.norm(grad)<self.tol: break

            # --- CALCOLO DIREZIONE ---
            if self.metodo=='gd':
                direzione=-grad
            elif self.metodo=='newton':
                H=self.calcola_hessiana(self.w,X_bias,ypred)
                direzione=np.linalg.solve(H,-grad)
            elif self.metodo=='cgn':
                B_k=self.calcola_hessiana(self.w,X_bias,ypred)
                direzione=self.cg_newton(grad,B_k)
            elif self.metodo=='bfgs':
                direzione=-np.dot(H_inv,grad)
            elif self.metodo=='lbfgs':
                direzione=self.lbfgs(grad)

            # --- LINE SEARCH ---
            if self.metodo in ['gd','newton','cgn']:
                alpha=self.backtracking_ls_armijo(self.w,X_bias,y,direzione,loss,grad,i,loss_precedente)
            elif self.metodo in ['bfgs','lbfgs']:
                alpha=self.ls_wolfe(self.w,X_bias,y,direzione,loss,grad)

            w_new=self.w+alpha*direzione

            # --- AGGIORNAMENTO BFGS - L-BFGS ---
            if self.metodo in ['bfgs','lbfgs']:
                _,grad_new,_=self.calcola_gradiente_loss(w_new,X_bias,y)
                s_k=w_new-self.w
                y_k=grad_new-grad
                rho_den=np.dot(y_k,s_k)

                if rho_den>1e-10:
                    if self.metodo=='bfgs':
                        #NOCEDAL: SCALING INIZIALE SOLO ALLA PRIMA ITERAZIONE ---
                        if i==0:
                            gamma_0=np.dot(s_k,y_k)/np.dot(y_k,y_k)
                            H_inv=H_inv*gamma_0 #scala la matrice identità iniziale
                        # ----------------------------------------------------------------------

                        rho_k=1.0/rho_den
                        I=np.eye(len(self.w))
                        A=I-rho_k*np.outer(s_k,y_k)
                        B=I-rho_k*np.outer(y_k,s_k)
                        H_inv=np.dot(A,np.dot(H_inv,B))+rho_k*np.outer(s_k,s_k)
                    elif self.metodo=='lbfgs':
                        self.s_history.append(s_k)
                        self.y_history.append(y_k)
                        if len(self.s_history)>self.m:
                            self.s_history.pop(0)
                            self.y_history.pop(0)

            self.w=w_new
            loss_precedente=loss

    def predict_proba(self,X):
        X_bias=np.insert(X,0,1,axis=1)
        return self.sigmoide(np.dot(X_bias,self.w)).flatten()

#-------------------------------------------------------------------
# 3. CROSS-VALIDATION (OTTIMIZZAZIONE LAMBDA)
# ------------------------------------------------------------------
metodi=['gd','newton','cgn','bfgs','lbfgs']
lambda_grid=[0.001,0.01,0.1,1,10]

skf=StratifiedKFold(n_splits=5,shuffle=True,random_state=42)
risultati_cv={m:{l:[] for l in lambda_grid} for m in metodi}

print("\n Grid Search (5-Fold CV) per ottimizzare Lambda")

for fold,(train_index,val_index) in enumerate(skf.split(X_train_full,y_train_full),1):
    X_train_fold,X_val_fold=X_train_full.iloc[train_index],X_train_full.iloc[val_index]
    y_train_fold,y_val_fold=y_train_full.iloc[train_index],y_train_full.iloc[val_index]

    X_train_processed=ct.fit_transform(X_train_fold)
    X_val_processed=ct.transform(X_val_fold)

    for l_val in lambda_grid:
        for m in metodi:
            modello=RegressioneLogistica(metodo=m,max_iter=1000,lambda_reg=l_val)
            modello.fit(X_train_processed,y_train_fold)

            y_prob_val=modello.predict_proba(X_val_processed)
            loss_val=-np.mean(y_val_fold*np.log(y_prob_val)+(1-y_val_fold)*np.log(1-y_prob_val))
            risultati_cv[m][l_val].append(loss_val)

migliori_parametri={}
print("\nRISULTATI GRID SEARCH")
for m in metodi:
    best_l=min(lambda_grid,key=lambda l:np.mean(risultati_cv[m][l]))
    best_loss=np.mean(risultati_cv[m][best_l])

    migliori_parametri[m]=best_l
    print(f"Metodo {m.upper():<18} MIGLIOR LAMBDA: {best_l:<5} (Loss: {best_loss:.4f})")

#-------------------------------------------------------------
# 4. TEST FINALE 
# ------------------------------------------------------------
print(" TEST FINALE SUL TEST SET ")
X_train_full_processed=ct.fit_transform(X_train_full)
X_test_processed=ct.transform(X_test)

modelli_ottimali={}
metriche_roc={}
soglia_fissa=0.5

for m in metodi:
    l_ottimale=migliori_parametri[m]

    modello=RegressioneLogistica(metodo=m,max_iter=4000,lambda_reg=l_ottimale)
    modello.fit(X_train_full_processed,y_train_full)
    modelli_ottimali[m]=modello

    y_prob=modello.predict_proba(X_test_processed)

    fpr,tpr,_=roc_curve(y_test,y_prob)
    roc_auc=auc(fpr,tpr)
    metriche_roc[m]=(fpr,tpr,roc_auc)

    y_pred=(y_prob>=soglia_fissa).astype(int)

    print(f"\n METODO: {m.upper()} | AUC: {roc_auc:.4f} ")
    print(classification_report(y_test,y_pred,target_names=['Sani (0)','Malati (1)']))

print("RIEPILOGO CONVERGENZA E LOSS SUL TRAINING SET")
for m in metodi:
    modello=modelli_ottimali[m]
    num_iterazioni=len(modello.loss_storia)
    loss_finale_train=modello.loss_storia[-1]
    print(f"{m.upper():<18} Iterazioni: {num_iterazioni:<4} | Train Loss Finale: {loss_finale_train:.6f}")

# -----------------------------------------
# 5. VISUALIZZAZIONE RISULTATI
# -------------------------------------------
colori_roc={
    'gd':'green',
    'newton':'blue',
    'cgn':'cyan',
    'bfgs':'purple',
    'lbfgs':'darkorange'
}

stili_linea={
    'gd':('-',2),
    'newton':('-',2),
    'cgn':('-',2),
    'bfgs':('-',2),
    'lbfgs':('-',2)
}

#1. ANDAMENTO DELLA LOSS 
plt.figure(figsize=(10,6))
for m in metodi:
    modello=modelli_ottimali[m]
    num_iterazioni=len(modello.loss_storia)
    etichetta=f"{m.upper()} ({num_iterazioni} iterazioni)"
    ls,lw=stili_linea[m]

    plt.plot(range(1,num_iterazioni+1),modello.loss_storia,color=colori_roc[m],
             linestyle=ls,label=etichetta,linewidth=lw,alpha=0.8)

plt.title("Andamento della Loss sul training set",fontsize=14,fontweight='bold')
plt.xlabel("Iterazioni",fontsize=12)
plt.ylabel("Loss",fontsize=12)
plt.xscale('log')
plt.xticks([1,10,100,1000,3000],['1','10','100','1000','3000'])
plt.legend(fontsize=10)
plt.grid(True,which="both",linestyle='--',alpha=0.5)
plt.tight_layout()
plt.show()

#2. CURVE ROC
plt.figure(figsize=(9,7))
for m in metodi:
    fpr,tpr,roc_auc=metriche_roc[m]
    ls,lw=stili_linea[m]
    plt.plot(fpr,tpr,color=colori_roc[m],linestyle=ls,lw=lw,alpha=0.8,label=f'{m.upper()} (AUC = {roc_auc:.3f})')

plt.plot([0,1],[0,1],color='gray',lw=1.2,linestyle='-',label='Classificatore Casuale')
plt.xlim([0.0,1.0]); plt.ylim([0.0,1.05])
plt.xlabel('Tasso Falsi Positivi (FPR)',fontsize=12)
plt.ylabel('Tasso Veri Positivi (TPR)',fontsize=12)
plt.title('Curva ROC sul Test Set',fontsize=15)
plt.legend(loc="lower right",fontsize=10)
plt.grid(alpha=0.4,linestyle='--')
plt.tight_layout()
plt.show()

#3. MATRICI DI CONFUSIONE
fig = plt.figure(figsize=(15, 9))
fig.suptitle('Matrici di Confusione sul test set', fontsize=16, fontweight='bold')

for i, m in enumerate(metodi):
    ax = fig.add_subplot(2, 3, i + 1)
    
    modello = modelli_ottimali[m]
    y_prob = modello.predict_proba(X_test_processed)
    y_pred = (y_prob >= soglia_fissa).astype(int)
    cm = confusion_matrix(y_test, y_pred)

    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax, cbar=False, annot_kws={"size": 16},
                xticklabels=['Sano', 'Malato'], yticklabels=['Sano', 'Malato'])

    ax.set_title(m.upper(), fontsize=14, fontweight='bold')
    ax.set_xlabel('Predetto', fontsize=12)
    ax.set_ylabel('Reale', fontsize=12)

plt.tight_layout()
plt.show()

# -----------------------------------------
# 6. IMPATTO DELLE FEATURE
#------------------------------------------
print("\nIMPATTO DELLE FEATURE")

#estrazione pesi(senza bias w0)
pesi=modelli_ottimali['bfgs'].w[1:]
nomi=[n.replace('num__','').replace('cat__','').replace('remainder__','') 
        for n in ct.get_feature_names_out()]

#creazione DataFrame
df_imp=pd.DataFrame({'Sintomo':nomi,'Peso':pesi})
df_imp['Importanza']=df_imp['Peso'].abs()
df_imp['Effetto']=np.where(df_imp['Peso']>0,'Fattore di Rischio (+)','Fattore Protettivo (-)')

#ordino dal più importante in cima al meno importante in fondo
df_imp=df_imp.sort_values(by='Importanza',ascending=False)

print(df_imp[['Sintomo','Peso']].to_string(index=False,float_format=lambda x:f"{x:.4f}"))

plt.figure(figsize=(10,8))
sns.barplot(data=df_imp,x='Peso',y='Sintomo',hue='Effetto',
            palette={'Fattore di Rischio (+)':'tab:red','Fattore Protettivo (-)':'tab:blue'})

plt.axvline(x=0,color='black',linewidth=1.2)
plt.title('Impatto delle feature',fontsize=15,fontweight='bold')
plt.xlabel('peso (w)',fontsize=12)
plt.ylabel('')
plt.grid(axis='x',linestyle='--',alpha=0.6)
plt.legend(loc='lower right',fontsize=11,framealpha=0.9)

plt.tight_layout()
plt.show()
```
